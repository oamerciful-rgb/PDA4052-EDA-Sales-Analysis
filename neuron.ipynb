```python
inputs = [2.0, 1.0, 3.0]
weights = [0.5, 0.4, -0.6]
bias = 1.0

output = (
    inputs[0]*weights[0] +
    inputs[1]*weights[1] +
    inputs[2]*weights[2] +
    bias
)

print('Output:', output)
```

Observation

Increasing the first weight increased the output significantly because the first input (2.0) now has more influence on the neuron.

```python
inputs = [1.0, 1.0, 1.0]
weights = [0.2, 0.8, -0.5]
bias = 0.5

output = (
    inputs[0]*weights[0] +
    inputs[1]*weights[1] +
    inputs[2]*weights[2] +
    bias
)

print('Output:', output)
```

Observation

The second input has the strongest effect because its weight (0.8) is the largest positive weight.

The third input decreases the output because its weight is negative (-0.5).

```python
inputs = [2.0, 3.0, 1.0]

weightsA = [0.5, 0.2, -0.1]
weightsB = [-0.4, 0.9, 0.3]

biasA = 1.0
biasB = 0.5

outputA = sum(i*w for i, w in zip(inputs, weightsA)) + biasA
outputB = sum(i*w for i, w in zip(inputs, weightsB)) + biasB

print('Neuron A output:', outputA)
print('Neuron B output:', outputB)
```

Observation

Both neurons receive the same inputs, but they produce different outputs because they use different weights and biases. Each neuron “focuses” on different parts of the input.

```python
import numpy as np

inputs = [
    [1.0, 2.0, 3.0],
    [2.0, 1.0, 0.5],
    [3.0, 2.5, 1.0]
]

weights = [0.4, -0.2, 0.6]
bias = 1.0

outputs = np.dot(inputs, weights) + bias

print('Batch outputs:', outputs)
```

Observation

Instead of processing one input at a time, NumPy processes all samples together. This is faster and more efficient, especially for large datasets in AI systems.

Part 5: Reflection Questions
1. Why is bias important in a neuron?

Bias allows the neuron to shift its output value. Without bias, the neuron would always pass through zero and would be less flexible when learning patterns.

2. Can two neurons give different outputs for the same input? Why?

Yes. Different neurons can use different weights and biases, so they interpret the same input differently and produce different outputs.

3. Why is batch processing useful in real-world AI systems?

Batch processing allows neural networks to handle many samples at once, making training and prediction much faster and more efficient using optimized matrix operations.
